```python
def safe_divide(num,deno):
    try:
        num=int(input('enter the numerator'))
        deno=int(input('enter the denominator'))
        a=num/deno
        return a
    except ZeroDivisionError:
        print('The number you provided is 0')
        return None
    else:
        print('success a=', a)
    finally:
        print('process completed')
safe_divide(10,5)
```
